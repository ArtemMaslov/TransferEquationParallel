Эталонная программа (5000x5000)

```python
import matplotlib.pyplot as plt
import numpy
```

```python
nprocs = numpy.array([2, 3, 4])
```

```python
time_seq = 0.2377
times_mpi = numpy.array([0.1308, 0.0936, 0.0785])
times_mp  = numpy.array([0.1322, 0.0865, 0.0649])

plt.title("speedup эталонная программа")
plt.plot(nprocs, time_seq / times_mpi, label="OpenMPI", marker='o')
plt.plot(nprocs, time_seq / times_mp, label="OpenMP", marker='o')
plt.plot(nprocs, nprocs, label ="ideal", marker='o')
plt.grid()
plt.legend()
plt.ylabel("speedup")
plt.xlabel("Nproc")
plt.show()
plt.clf()


plt.title("efficiency эталонная программа")
plt.plot(nprocs, 100 * (time_seq / times_mpi) / nprocs, label="OpenMPI", marker='o')
plt.plot(nprocs, 100 * (time_seq / times_mp) / nprocs, label="OpenMP", marker='o')
plt.grid()
plt.legend()
plt.ylabel("efficiency")
plt.xlabel("Nproc")
plt.show()
plt.clf()
```
